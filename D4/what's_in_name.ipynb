## Dynamic Approach (Alphabetical Position)

If you want to map any name based on its actual position in the English alphabet (where A=1, B=2, C=3, D=4, etc.):

```python
name = input("Enter your name:")
print(f"Hello {name}")

total_sum=0     # <<-------------------------------don't forget
for ARUN in name.upper():
    # ord(letter) - 64 converts ASCII values ('A'=65) to alphabet positions (1 to 26)
    position = ord(ARUN) - 64
    print(f"{ARUN} = {position}")
    total_sum += position

# Extra line 2: Print the final total sum calculation
print(f"Total_Sum = {total_sum}")
```

```python
def letter_value(ch):
    """Return the alphabet position (A=1 ... Z=26) for a single letter."""
    ch = ch.upper()
    if 'A' <= ch <= 'Z':
        return ord(ch) - ord('A') + 1
    return None  # for spaces, punctuation, etc.

def name_to_values(name):
    """Return a list of (letter, value) pairs for each alphabetic character in the name."""
    result = []
    for ch in name:
        val = letter_value(ch)
        if val is not None:
            result.append((ch.upper(), val))
    return result

def total_value(name):
    """Sum of all letter values in the name."""
    return sum(v for _, v in name_to_values(name))

def reduce_to_single_digit(n):
    """Reduce a number to a single digit by repeated digit-summing (common in numerology)."""
    while n > 9:
        n = sum(int(d) for d in str(n))
    return n


if __name__ == "__main__":
    name = input("Enter a name: ")

    pairs = name_to_values(name)
    print("\nLetter\tValue")
    for letter, val in pairs:
        print(f"{letter}\t{val}")

    total = total_value(name)
    print(f"\nTotal value: {total}")
    print(f"Reduced to single digit: {reduce_to_single_digit(total)}")
```

```python
# Python3 code to demonstrate working of
# Life Path Number
# Using mathematical formulas

# initializing string
# test_str = "19970314"

test_str = input("Enter your birthdate without space:")
# print(f"Hello {name}")

# printing original string
print("The original string is : " + str(test_str))

# using mathematical formulas
while len(test_str) > 1:
    sum = 0
    for digit in test_str:
        sum += int(digit)
    test_str = str(sum)

# printing result
print("Life Path Number : " + str(test_str))
```

```python
# Python3 code to demonstrate working of
# Life Path Number
# Using loop

# initializing string
test_str = "19970314"

# printing original string
print("The original string is : " + str(test_str))

res = 0
for num in test_str:
    res += int(num)
    
    # modulation in case of 2 digit number
    if res > 9:
        res = res % 10 + res // 10

# printing result
print("Life Path Number  : " + str(res))
```
